```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style="whitegrid")

print("Libraries imported successfully!")
```

```python
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9, 1, 10],
    'Sleep_Hours': [8, 7, 7, 6, 8, 6, 7, 5, 9, 6],
    'Attendance': [60, 65, 70, 75, 80, 85, 90, 95, 50, 98],
    'Marks': [45, 50, 55, 65, 70, 75, 85, 90, 40, 95]
}

df = pd.DataFrame(data)

df
```

```python
correlation_matrix = df.corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='Study_Hours',
    y='Marks'
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='Attendance',
    y='Marks',
    color='green'
)

plt.title("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='Sleep_Hours',
    y='Marks',
    color='orange'
)

plt.title("Sleep Hours vs Marks")
plt.xlabel("Sleep Hours")
plt.ylabel("Marks")

plt.show()
```

```python
print("Strong Positive Correlations:")

print(
    correlation_matrix[
        (correlation_matrix > 0.7) &
        (correlation_matrix < 1.0)
    ].stack()
)

print("\nStrong Negative Correlations:")

print(
    correlation_matrix[
        correlation_matrix < -0.7
    ].stack()
)
```
